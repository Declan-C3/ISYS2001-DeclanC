## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Can you help me implement a savings goal tracker that requires input from the user, and gives different feedback depending on the user's current progress which will be based on differing factors and decision logic, with the code using if/elif/else operators.

Factor 1: Current Savings Amount Factor 2: Current Savings Goal Factor 3: Feedback Given

Decision Logic: User will be prompted to enter initial savings amount, and their savings goal amount. Once inputted, a range of feedback and statements will be provided to the user depending on the current amount and desired amount.

User will also be prompted with a time component on when they want to reach their savings goal, and be provided with how much they would need to save monthly to reach their goal.

Additionally, Users will be prompted with specified advice on how to reach their savings goal based on their current_savings amount. This should be measured in 25% increments with differing advice and messages.

At >= 0%, users should be encouraged and given advice in setting up automatic transfers of X amount of dollars each payday and look to find ways to cut down expenses.

At >= 25%, users should be advised to set up automatic transfers of X amount of dollars each payday.

At >= 50%, users should be advised to look for ways to cut down expenses to reach their goal

At >= 75%, users should be encourage that they are nearly there.

Ensure input valdiation through numerical inputs for current savings, savings goal, and months to goal.

```

````python
def savings_tracker():
    print("Welcome to your Savings Goal Tracker!")

    while True:
        try:
            current_savings_str = input("Enter your current savings amount: ")
            current_savings = float(current_savings_str)
            if current_savings < 0:
                print("Current savings cannot be negative. Please enter a non-negative number.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a numerical value for current savings.")

    while True:
        try:
            savings_goal_str = input("Enter your savings goal amount: ")
            savings_goal = float(savings_goal_str)
            if savings_goal <= 0:
                print("Savings goal must be a positive number. Please enter a positive value.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a numerical value for savings goal.")

    # Prompt for time component if goal is not yet met
    months_to_goal = 0
    if current_savings < savings_goal:
        while True:
            try:
                months_str = input("In how many months do you want to reach your goal? ")
                months_to_goal = int(months_str)
                if months_to_goal <= 0:
                    print("Number of months must be a positive integer. Please enter a positive value.")
                    continue
                break
            except ValueError:
                print("Invalid input. Please enter an integer for the number of months.")

    print(f"\nYou currently have ${current_savings:.2f} towards your goal of ${savings_goal:.2f}.")

    if current_savings >= savings_goal:
        print("Congratulations! You have reached and even surpassed your savings goal! 🎉")
    else:
        progress_percentage = (current_savings / savings_goal) * 100
        print(f"You have achieved {progress_percentage:.2f}% of your goal.")

        remaining_to_save = savings_goal - current_savings
        monthly_saving_needed = 0
        if months_to_goal > 0:
            monthly_saving_needed = remaining_to_save / months_to_goal
            print(f"To reach your goal in {months_to_goal} months, you would need to save approximately ${monthly_saving_needed:.2f} per month.")
        else:
            print("You still need to save more to reach your goal.")

        # Specific advice based on 25% increments
        if progress_percentage >= 75:
            print("You're nearly there! Keep pushing, you're almost at your goal! 💪")
        elif progress_percentage >= 50:
            print("Great progress! You're more than halfway to your goal. Look for ways to cut down expenses to reach your goal. ✨")
        elif progress_percentage >= 25:
            print(f"Good start! You're making progress. Consider setting up automatic transfers of a fixed amount each payday to stay consistent. (e.g., if you need to save ${monthly_saving_needed:.2f} monthly, divide it by your number of paydays in a month). 👍")
        else: # progress_percentage < 25 (covers >= 0%)
            print(f"Every penny counts! To kickstart your savings, consider setting up automatic transfers of a fixed amount each payday (e.g., if you need to save ${monthly_saving_needed:.2f} monthly, divide it by your number of paydays in a month) and actively look for ways to cut down expenses. You can do it! 🚀")

# Run the tracker
savings_tracker()
````
