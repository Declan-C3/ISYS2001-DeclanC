### Activity 5: Build an admissions decision system

Universities weigh several factors to make admission decisions. Design and build a fair, logical version.

**Design first:**

1. Choose 3 to 4 factors (GPA, test scores, activities, essay, special circumstances)
2. Decide how they combine (all required? at least 2 of 3?)
3. Set specific thresholds
4. Think about edge cases and tie-breakers

> **Your criteria** (double-click to edit):

```
Factor 1: Special Circumstances
Factor 2: Extracurricular Activities
Factor 3: GPA
Decision logic: GPA should be the main factor as meeting a certain GPA threshold should allow admission, in case of tie-breakers or too many student allocations, decisions should factor in special circumstances or extracurricular activities that the students may have been involved in.

A tiered GPA threshold will determine admittance with ranges of >= 3.5 will automatically admit students, >= 3.0 will place students on a list and will grant students allocations based on positioning and other factors, >= 2.0 will require special circumstances or extracurricular activities to be admitted.

Additionally, if students don't meet GPA threshold, checks should be done to factor in their special circumstances and/or extracurricular acitivities to allow students who don't reach the GPA threshold and grant admittance.

Special Circumstances and Extracurricular Activities will be based on a qualitative level determing how strong these other factors are to accomodate a student's application.
```

````python
# Your admissions decision system

def get_admission_decision(
    gpa: float,
    special_circumstances_level: str, # e.g., "Strong", "Moderate", "Weak", "None"
    extracurricular_level: str       # e.g., "Strong", "Moderate", "Weak", "None"
) -> str:
    """
    Determines university admission status based on GPA, special circumstances,
    and extracurricular activities.

    Args:
        gpa: The applicant's Grade Point Average (0.0-4.0).
        special_circumstances_level: Qualitative level of special circumstances.
        extracurricular_level: Qualitative level of extracurricular activities.

    Returns:
        A string representing the admission decision.
    """

    # Convert qualitative levels to a numeric score for easier comparison
    # You can refine these scores based on your exact weighting
    level_scores = {"Strong": 3, "Moderate": 2, "Weak": 1, "None": 0}
    sc_score = level_scores.get(special_circumstances_level, 0)
    ec_score = level_scores.get(extracurricular_level, 0)

    if gpa >= 3.5:
        return "Automatic Admission"
    elif gpa >= 3.0:  # GPA between 3.0 and 3.49
        # "place students on a list and will grant students allocations based on positioning and other factors"
        # We'll consider 'Moderate' or 'Strong' in either SC or EC as a positive factor
        if sc_score >= 2 or ec_score >= 2: # At least Moderate in either
            return "Provisional Admission (Review for Final Allocation)"
        else:
            return "Waitlisted (Needs Further Review)"
    elif gpa >= 2.0:  # GPA between 2.0 and 2.99
        # "will require special circumstances or extracurricular activities to be admitted"
        # This implies a significant impact from these factors is needed.
        if sc_score >= 2 or ec_score >= 2: # At least Moderate in either
            return "Conditional Admission (Strong Non-GPA Factors)"
        else:
            return "Rejected (Insufficient GPA & Non-GPA Factors)"
    else:  # GPA below 2.0
        return "Rejected (GPA Below Minimum Threshold)"

# Example Test Cases:
print("--- Admission Decision Examples ---")

# Student 1: High GPA
student_1_gpa = 3.8
student_1_sc = "None"
student_1_ec = "Weak"
print(f"Student 1 (GPA: {student_1_gpa}, SC: {student_1_sc}, EC: {student_1_ec}): "
      f"{get_admission_decision(student_1_gpa, student_1_sc, student_1_ec)}")

# Student 2: Borderline GPA (3.0-3.49) with strong EC
student_2_gpa = 3.2
student_2_sc = "Weak"
student_2_ec = "Strong"
print(f"Student 2 (GPA: {student_2_gpa}, SC: {student_2_sc}, EC: {student_2_ec}): "
      f"{get_admission_decision(student_2_gpa, student_2_sc, student_2_ec)}")

# Student 3: Borderline GPA (3.0-3.49) with no strong non-GPA factors
student_3_gpa = 3.1
student_3_sc = "Weak"
student_3_ec = "Weak"
print(f"Student 3 (GPA: {student_3_gpa}, SC: {student_3_sc}, EC: {student_3_ec}): "
      f"{get_admission_decision(student_3_gpa, student_3_sc, student_3_ec)}")

# Student 4: Lower GPA (2.0-2.99) with strong SC
student_4_gpa = 2.5
student_4_sc = "Strong"
student_4_ec = "Moderate"
print(f"Student 4 (GPA: {student_4_gpa}, SC: {student_4_sc}, EC: {student_4_ec}): "
      f"{get_admission_decision(student_4_gpa, student_4_sc, student_4_ec)}")

# Student 5: Lower GPA (2.0-2.99) with weak non-GPA factors
student_5_gpa = 2.3
student_5_sc = "None"
student_5_ec = "Weak"
print(f"Student 5 (GPA: {student_5_gpa}, SC: {student_5_sc}, EC: {student_5_ec}): "
      f"{get_admission_decision(student_5_gpa, student_5_sc, student_5_ec)}")

# Student 6: Below minimum GPA
student_6_gpa = 1.9
student_6_sc = "Strong"
student_6_ec = "Strong"
print(f"Student 6 (GPA: {student_6_gpa}, SC: {student_6_sc}, EC: {student_6_ec}): "
      f"{get_admission_decision(student_6_gpa, student_6_sc, student_6_ec)}")
````

### Activity 6: Debug the tricky one

A student wrote this scholarship checker. The rule they wanted:

> Eligible if GPA is 3.0 or higher, AND (they have 50+ community service hours OR financial need), AND the essay scored 7 or higher.

Run it and test it. It does not behave the way they intended.

````python
print("=== Scholarship Eligibility Checker ===")

gpa = float(input("Enter GPA (0.0-4.0): "))
community_hours = int(input("Community service hours: "))
financial_need = input("Have financial need? (yes/no): ").lower()
essay_score = int(input("Essay score (1-10): "))

# Intended rule:
# GPA 3.0+ AND (community hours 50+ OR financial need) AND essay 7+

if gpa >= 3.0 and community_hours >= 50 or financial_need == "yes" and essay_score >= 7:
    print("Eligible for scholarship!")
else:
    print("Not eligible")
````

**Work through it.**

- Write the intended rule in plain English (you have it above).
- Python evaluates `and` before `or`. So how does the computer actually group that condition? Add brackets to show the grouping Python is really using.
- Where does that grouping differ from what the student meant?

> **Your analysis** (double-click to edit):

```
What Python actually checks:
if (gpa >= 3.0 and community_hours >= 50) or (financial_need == "yes" and essay_score >= 7):

Why that is wrong:
This is wrong because it determines either side of the or operations as two separate conditions for the scholarship eligibility, meaning that if one side is eligble then the student overall is eligble. This isn't how it's supposed to work as gpa, and the essay score should both be considered essential and the community hours and financial need are the two checks that only one are needed for eligibilty.

The fix:
Using parathesis between the community hours and financial need checks to ensure that the prompt runs correctly.
```

Write your corrected version below.

````python
# Your corrected scholarship checker here
print("=== Scholarship Eligibility Checker ===")

gpa = float(input("Enter GPA (0.0-4.0): "))
community_hours = int(input("Community service hours: "))
financial_need = input("Have financial need? (yes/no): ").lower()
essay_score = int(input("Essay score (1-10): "))

# Intended rule:
# GPA 3.0+ AND (community hours 50+ OR financial need) AND essay 7+

if gpa >= 3.0 and (community_hours >= 50 or financial_need == "yes") and essay_score >= 7:
    print("Eligible for scholarship!")
else:
    print("Not eligible")
````

> "This should check (A AND (B OR C) AND D) but it is not working. Can you explain how operator precedence is grouping my condition, and how to fix it?"

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
- Give it your intent when you ask for help, especially when debugging.
